# Exploratory analysis

Quick look at a dataset: class balance and the mean pixel-intensity
distribution that drives intensity filtering.

Run `python tests/make_synthetic_data.py data/synthetic` first if you don't
have real data prepared yet.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt

from mammokit.data.dicom_reader import read_mammogram_kw
from mammokit.data.filtering import mean_intensity
```

```python
df = pd.read_csv('../data/synthetic/annotations.csv')
display(df['label'].value_counts())
df.head()
```

```python
mi = mean_intensity(df, '../data/synthetic', read_mammogram_kw)
plt.hist(mi, bins=30)
plt.yscale('log'); plt.xlabel('mean intensity'); plt.ylabel('count (log)')
plt.show()
```
